TSS-centered ATAC-seq heatmaps revealed strong enrichment of chromatin accessibility around transcription start sites in both cDC1 and cDC2, demonstrating high-quality ATAC-seq signal and promoter-proximal accessibility changes. When stratified by gain and loss peaks, cDC1 samples showed relatively similar accessibility patterns between WT and KO, indicating limited global chromatin remodeling in this population. In contrast, cDC2 samples exhibited higher overall signal intensity and greater heterogeneity, particularly in KO conditions, suggesting more pronounced accessibility changes. These trends are consistent with the original study’s Figure 6A–B, which reported that HDAC1 deletion has a stronger impact on chromatin accessibility in cDC2 than in cDC1
Compared to the original Figure 6A–B, the TSS-centered ATAC-seq heatmaps reproduced here show similar promoter-focused enrichment patterns but reduced contrast between gain and loss peak groups. In the original study, gain and loss peaks display more pronounced differences in signal intensity, particularly in cDC2, whereas our heatmaps exhibit more uniform accessibility profiles across categories. This discrepancy likely reflects differences in peak selection and filtering strategies, as the original analysis applied stricter thresholds and integrated additional epigenetic information to refine regulatory regions. Moreover, the absence of RNA-seq and H3K27ac integration in our analysis limits the ability to stratify peaks by functional relevance, resulting in less separation between accessibility gain and loss. Despite these differences, the overall TSS enrichment and stronger accessibility changes in cDC2 relative to cDC1 are consistent with the trends reported in the original figure.

The original Figure 6D presents genome viewer tracks to illustrate a representative cDC1-specific gain of chromatin accessibility at the Maged1 locus, but does not explicitly specify the file format used for visualization. In this reproduction, ATAC-seq accessibility at the same locus was visualized using normalized bigWig coverage files rather than read-level pileups. Consequently, the reproduced tracks display smoother signal profiles and reduced local variability. Despite these differences in visualization, increased chromatin accessibility in cDC1 knockout cells relative to wild-type is clearly observed at the Maged1 locus, consistent with the biological trend reported in the original study.


6F

Figure shows an example of a cDC2-specific gain of chromatin accessibility at the SpiB locus. Consistent with the original study, ATAC-seq signal is increased in cDC2 knockout cells compared to wild-type, whereas changes in cDC1 are relatively modest. Due to data availability, only ATAC-seq bigWig tracks were visualized in this reproduction, and Cut&Run (H3K27ac) profiles shown in the original figure were not included. Nevertheless, the observed increase in accessibility at the SpiB locus in cDC2 supports the same biological conclusion reported in the original study.
